```python
import pandas as pd
import sqlite3

conn = sqlite3.connect("../data/processed/olist.db")
```

```python
products = pd.read_sql("SELECT * FROM products;", conn)
products = products.dropna(subset=["product_weight_g", "product_length_cm", "product_height_cm", "product_width_cm"])
products["product_category_name"] = products["product_category_name"].fillna("unknown")
products.to_sql("products", conn, if_exists="replace", index=False)
print("Products cleaned:", products.shape)
```

```python
reviews = pd.read_sql("SELECT * FROM order_reviews;", conn)
reviews["review_comment_title"] = reviews["review_comment_title"].fillna("")
reviews["review_comment_message"] = reviews["review_comment_message"].fillna("")
reviews.to_sql("order_reviews", conn, if_exists="replace", index=False)
print("Reviews cleaned:", reviews.shape)
```

```python
orders = pd.read_sql("SELECT * FROM orders;", conn)
date_cols = ["order_purchase_timestamp", "order_approved_at", "order_delivered_carrier_date",
             "order_delivered_customer_date", "order_estimated_delivery_date"]
for col in date_cols:
    orders[col] = pd.to_datetime(orders[col])
orders.to_sql("orders", conn, if_exists="replace", index=False)
print("Orders cleaned:", orders.shape)
```
